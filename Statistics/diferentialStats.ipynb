```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
data = {
    'Income': [25000, 28000, 30000, 29000, 27000, 100000, 26000],
    'Car_Color': ['Red', 'Red', 'Blue', 'Black', 'Red', 'Blue', 'Black']
}
df=pd.DataFrame(data)
```

```python
df.head()
```

```python
mean_income=df['Income'].mean()
print('The average income:-',mean_income)
```

```python
median_income=df['Income'].median()
print("The median is:-",median_income)
```

```python
mode_sale=df['Car_Color'].mode()[0]
print(mode_sale)
```

## Standard Deviation and variance

```python
df['Income'].std()
```

```python
df['Income'].var()
```

## Range

```python
range_income=df['Income'].max() - df['Income'].min()
range_income
```

## Percentile and quartiles

```python
q1 = df['Income'].quantile(0.25)
q2 = df['Income'].quantile(0.50)
q3 = df['Income'].quantile(0.75)
print("Q1 (25%):", q1)
print("Q2 (50% - Median):", q2)
print("Q3 (75%):", q3)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df['Income'])
plt.title('Income Distribution')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df['Income'],kde=True)
plt.title('Income Distribution')
plt.show()
```
