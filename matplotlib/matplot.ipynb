```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig,ax=plt.subplots()
ax.set_title("Basic graph")
ax.set_xlabel('Days')
ax.set_ylabel('Total Patient')
plt.show()

```

```python
#LINE PLOT 

months=['JAN','FEB','MAR','APR','MAY','JUN','JUL']
recovery_rate=[55,75,89,65,36,45,78]
plt.plot(months,recovery_rate,marker='o',linestyle='-',color='b',label='Recovery Rate')
plt.xlabel('Months')
plt.ylabel('Recovery rate')
plt.title("Recover rate analysis")
plt.legend()
plt.show()
```

```python
#Bar Plot
medicines=['Paracetemol','Insulin','Painkiller','Antacids','AntiBiotics','AntiAging']
demand=[65,55,75,85,58,86]
plt.bar(medicines,demand,color=['r','g','blue','purple','black','orange'],label="Demand")
plt.xlabel("Medicines")
plt.ylabel("Demand")
plt.title("Medicines Trend")
plt.legend()
plt.show()
```

```python
##pie chart
Diseases=['Rabies','Dengue','Malaria','Typhoid','Fever']
percentage=[5,25,30,27,59]
colors=['r','g','blue','purple','pink','orange']
plt.pie(percentage,labels=Diseases,autopct="%1.1f%%",colors=colors)
plt.title('Disease Distribution in city')
plt.show()
```

```python
#Scatter plot
time_spent=[2,4,3,5,6]
accuracy=[90,68,57,85,78]
plt.scatter(time_spent,accuracy,color='red')
plt.xlabel('Time spent by Doctors(hours)')
plt.ylabel("Diagnosis accuracy(%)")
plt.title("Time vs Diagonis Accuracy")
plt.show()
```

```python
##histogram
ages=np.random.randint(20,80,100)

plt.hist(ages,bins=10,color='blue',edgecolor='black')
plt.xlabel("Age Group")
plt.ylabel("No of patient")
plt.title("No of patient according to different age groups")
plt.show()
```
