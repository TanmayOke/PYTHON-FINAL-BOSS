```python
array_five=np.arange(1,26,1).reshape(5,5)
print(array_five)
```

```python
row_sum=np.sum(array_five,axis=1)
print(row_sum)
```

```python
col_sum=np.sum(array_five,axis=0)
print(col_sum)
```

```python
import numpy as np  

# Monthly claim amounts (in INR) for different policyholders
claims = np.array([
    [12000, 15000, 18000],  # Policyholder 1
    [10000, 13000, 17000],  # Policyholder 2
    [9000,  11000, 16000]   # Policyholder 3
])

print("Insurance Claim Data (INR):")
print(claims)
```

```python
print("The first insurance claim of policyholder1 in month 1:-",claims[0][0])
print("The second insurance claim of policyholder1 in month 1:-",claims[0][1])
print("The Third insurance claim of policyholder3 in month 3:-",claims[2][2])
```

```python
print("The total claim made by policyholders in first month:-",np.sum(claims,axis=1))
```

```python
print("The total claim made by policyholders in first month:-",np.sum(claims,axis=0))

```
