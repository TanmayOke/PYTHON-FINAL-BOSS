```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
sns.set(style="whitegrid")
plt.figure(figsize=(8,8))
```

```python
df=pd.read_csv("CAR_DETAILS_FROM_CAR_DEKHO.csv")
df.head()
```

```python
print("Fuel Type Distribution")
sns.countplot(data=df,x="fuel",hue="fuel",legend=False)
plt.xlabel("Type of Fuel")
plt.xlabel("Number of vehicle")
plt.title("Fuel Distribution")
plt.show()
```

```python
#INFERENCE=Diesel and petrol cars are more than any other fuel
```

```python
#Price Range Analysis
sns.histplot(df['selling_price'],bins=30,kde=True)
plt.xlabel("Selling Price(lakhs)")
plt.ylabel("Frequency")
plt.title("Selling Price Distribution")
plt.show()
```

```python
sns.boxplot(data=df,x="fuel",y="selling_price",palette="Set2")
plt.title("Price Distribution over fuel")
```

```python
sns.boxplot(data=df,x="transmission",y="selling_price",hue="transmission")
plt.title("Transmission sales")
plt.show()
```

```python
sns.violinplot(data=df,x="owner",hue="owner",y="selling_price")
plt.title("Owner Wise Distribution")
plt.show()
```
