```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("telecom_customer_data.csv")
df.head()
```

```python
#creating a scatter plot
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Call_Duration",y="Satisfaction_Score",hue="Network_Provider")
#moving legend out of box
plt.legend(title="Network Provider",bbox_to_anchor=(1,1),loc="upper left")


#Add labels and titles
plt.xlabel("Call Duration")
plt.ylabel("Satisfaction Score")
plt.title("Call Analysis")
plt.show()

```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df,x="Network_Provider",hue="Network_Provider",palette="Set2")
plt.title("Customer Distribution by network provider")
plt.xlabel("Network Provider")
plt.ylabel("count of customer")
plt.show()
```

```python
plt.figure(figsize=(7,6))
sns.histplot(df['Call_Duration'],bins=20,kde=True,color="blue")
plt.title("Call Duration Analysis")
plt.xlabel("Call Duration(minutes)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
number_values_df=df.select_dtypes(['number'])
sns.heatmap(number_values_df.corr(),annot=True,cmap="coolwarm")
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Network_Provider",
    y="Satisfaction_Score"
)

plt.title("Satisfaction Score by Network Provider")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="Network_Provider",
    y="Satisfaction_Score"
)

plt.title("Satisfaction Score Distribution by Provider")
plt.xlabel("Network Provider")
plt.ylabel("Satisfaction Score")

plt.show()
```
