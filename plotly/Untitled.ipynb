```python
import plotly.express as px 
import pandas as pd 
```

```python
df=pd.read_csv("tourism_data.csv")
df.head()
```

```python
fig=px.line(df,x='Month',y='Tourist_Arrivals',title="Ovearll tourism_data.csv")
fig.show()
```

```python
fig=px.bar(df,x='Country',y='Tourist_Arrivals',title='World wide tourism')
fig.show()
```

```python
df.head()
```

```python
fig=px.scatter(df,x='Average_Travel_Cost',y='Hotel_Bookings',title='Average Travel cost Vs Hotel Booking')
fig.show()
```

```python
fig=px.histogram(df,x='Hotel_Price',title="Distribution of Hotel Price")
fig.show()
```

```python
df.head()
```

```python
fig=px.violin(df,x='Season',y='Average_Travel_Cost',box=True,title='Season wise Tourism data')
fig.show()
```

```python
fig=px.timeline(df,x_start='Start_Date',x_end='End_Date' ,y='Activity',title='tourism')
fig.show()
```
