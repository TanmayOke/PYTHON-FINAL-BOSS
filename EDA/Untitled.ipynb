```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
sns.boxplot(y=df['Ozone'])
plt.title("Finding Ozone outlier")
```

```python
## Inter quantile range 
Q1=df['Ozone'].quantile(0.25)
Q3=df['Ozone'].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1 - 1.5 * IQR
upper_bound=Q3 + 1.5 * IQR
## Finding Outlier
outlier=df[(df['Ozone']<lower_bound)|(df['Ozone']>upper_bound)]
print("Number of outliers in Ozone cloumn:- ",outlier.shape)
```

## Now ozone column should be filled with median

```python
ozone_medain=df['Ozone'].median()
df.fillna({'Ozone':ozone_medain},inplace=True)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df['Solar'])
plt.title("Finding Solar outlier")
```

```python
solar_mean=df['Solar'].mean()
df.fillna({'Solar':solar_mean},inplace=True)
```

```python
weather_mode=df['Weather'].mode()[0]
df.fillna({'Weather':weather_mode},inplace=True)
```
