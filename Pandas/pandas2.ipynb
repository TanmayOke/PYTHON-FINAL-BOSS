```python
import pandas as pd
df=pd.read_csv('data_clean.csv')
df.head()
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull())
```

```python
median_ozone=df['Ozone'].median()
median_ozone
```

```python
#df['Ozone'].fillna(median_ozone,inplace=True)
df.fillna({'Ozone': median_ozone}, inplace=True)
```

```python
df.head()
```

```python
df['Solar.R'].isnull().sum()
```

```python
median_solar=df['Solar.R'].median()
df.fillna({'Solar.R':median_solar},inplace=True)
```

```python
df['Weather'].isna().sum()
```

```python
weather_median=df['Weather'].mode()
df.fillna({'Weather':weather_median},inplace=True)
```

```python
df['Temp'].isna().sum()
```

```python
df.head(10)
```

```python
weather_group=df.groupby('Weather')
print("The Average temperature on differnt weathers:-",weather_group['Temp'].mean())
```

```python
month_group=df.groupby('Month')
print("The maximum speed of wind in each month:-\n",month_group['Wind'].max())
```

```python
#basically groupby function ka kaam ye hai ki kisi ek column ka gropu bnakr usko dusre colum ke hisaab se data analysis kr ske
```

```python
print("The Value of ozone in each month:-\n",month_group['Ozone'].median())
```
