```python
entries_1 = {'swift': 649000,
           'santro': 391000,
           'audi': 5476000,
           'elantra': 1786000,
           'bolero': 1091000}
series_1=pd.Series(entries_1,name="price")
print(series_1)
```

```python
series_1.loc['swift']
```

```python
print(series_1.loc[['swift']])
print(series_1.loc[['audi']])
```

```python

series_1.loc[['swift','elantra']]
```

```python
#Slicing
series_1.loc['swift':'elantra']
```
